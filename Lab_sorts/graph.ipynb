```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
mass1 = np.array(pd.read_csv("1.csv"))
print (mass1)
N = []
bub = []
ins = []
sel = []

for i in range (len(mass1)):
    N.append(int(mass1[i][0]))
    bub.append(float(mass1[i][1])/5)
    ins.append(float(mass1[i][2])/5)
    sel.append(float(mass1[i][3])/5)

N = np.array(N)
bub = np.array(bub)
ins = np.array(ins)
sel = np.array(sel)
```

```python
coef_bub = np.polyfit(N, bub, 2)
coef_ins = np.polyfit(N, ins, 2)
coef_sel = np.polyfit(N, sel, 2)

x = []

for i in range (100, 100000, 100):
    x.append(i)

x = np.array(x)
y_bub = np.array(coef_bub[0]*x*x+coef_bub[1]*x+coef_bub[2])
y_ins = np.array(coef_ins[0]*x*x+coef_ins[1]*x+coef_ins[2])
y_sel = np.array(coef_sel[0]*x*x+coef_sel[1]*x+coef_sel[2])

plt.figure(figsize=(10, 6))

plt.scatter(N, bub, s=30)
plt.scatter(N, ins, s=30)
plt.scatter(N, sel, s=30)

plt.plot(x, y_bub, label=f'Buble_sort: t={coef_bub[0]: .2}*N^2{coef_bub[1]: .2}*N+{coef_bub[2]: .2}' )
plt.plot(x, y_ins, label=f'Insert_sort: t={coef_ins[0]: .2}*N^2{coef_ins[1]: .2}*N+{coef_ins[2]: .2}')
plt.plot(x, y_sel, label=f'Select_sort: t={coef_sel[0]: .2}*N^2{coef_sel[1]: .2}*N+{coef_sel[2]: .2}')

plt.xlabel('Number of items')
plt.ylabel('Time of sorts')
plt.grid()
plt.legend()
plt.savefig('Slow_sorts.pdf', format='pdf')
plt.show()
```

```python
log_N = np.log (N[2:])
log_bub= np.log (bub[2:])
log_ins = np.log (ins[2:])
log_sel = np.log (sel[2:])

plt.figure(figsize=(10, 6))

plt.scatter(log_N, log_bub, s=30)
plt.scatter(log_N, log_ins, s=30)
plt.scatter(log_N, log_sel, s=30)

plt.plot(log_N, log_bub, label='Buble_sort')
plt.plot(log_N, log_ins, label='Insert_sort')
plt.plot(log_N, log_sel, label='Select_sort')



plt.xlabel('log(Number of items)')
plt.ylabel('log(Time of sorts)')
plt.grid()
plt.legend()
plt.savefig('Slow_sorts_log.pdf', format='pdf')
plt.show()
```

```python
mass5 = np.array(pd.read_csv("5.csv"))

n = []
mer = []
shell = []
qui = []

for i in range (len(mass5)):
    n.append(int(mass5[i][0]))
    mer.append(float(mass5[i][1]))
    shell.append(float(mass5[i][2]))
    qui.append(float(mass5[i][3]))

n = np.array(n)
mer = np.array(mer)
shell = np.array(shell)
qui = np.array(qui)
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(n, mer, s=30, label='Merge_sort')
plt.scatter(n, shell, s=30, label='Shell_sort')
plt.scatter(n, qui, s=30, label='Quick_sort')


plt.xlabel('Number of items')
plt.ylabel('Time of sorts')
plt.grid()
plt.legend()
plt.savefig('Fast_sorts.pdf', format='pdf')
plt.show()
```

```python
mer_log = np.array(mer/(n*np.log(n)))
shell_log = np.array(shell/(n*np.log(n)))
qui_log = np.array(qui/(n*np.log(n)))

plt.figure(figsize=(10, 6))

plt.scatter(n, mer_log, s=30, label='Merge_sort')
plt.scatter(n, shell_log, s=30, label='Shell_sort')
plt.scatter(n , qui_log, s=30, label='Quick_sort')


plt.xlabel('Number of items')
plt.ylabel('Time of sorts/(N*log(N))')
plt.grid()
plt.legend()
plt.savefig('Fast_sorts_proof.pdf', format='pdf')
plt.show()
```
