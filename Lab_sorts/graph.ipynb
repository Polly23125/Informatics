```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from random import randint
from time import time
```

```python
mass1 = np.array(pd.read_csv("1.csv"))
print (mass1)
N = []
bub = []
ins = []
sel = []

for i in range (len(mass1)):
    N.append(int(mass1[i][0]))
    bub.append(float(mass1[i][1]))
    ins.append(float(mass1[i][2]))
    sel.append(float(mass1[i][3]))

N = np.array(N)
bub = np.array(bub)
ins = np.array(ins)
sel = np.array(sel)
```

```python
coef_bub = np.polyfit(N, bub, 2)
coef_ins = np.polyfit(N, ins, 2)
coef_sel = np.polyfit(N, sel, 2)

x = []

for i in range (100, 100000, 100):
    x.append(i)

x = np.array(x)
y_bub = np.array(coef_bub[0]*x*x+coef_bub[1]*x+coef_bub[2])
y_ins = np.array(coef_ins[0]*x*x+coef_ins[1]*x+coef_ins[2])
y_sel = np.array(coef_sel[0]*x*x+coef_sel[1]*x+coef_sel[2])

plt.figure(figsize=(10, 6))

plt.scatter(N, bub, s=30)
plt.scatter(N, ins, s=30)
plt.scatter(N, sel, s=30)

plt.plot(x, y_bub, label=f'Buble_sort: t={coef_bub[0]: .2}*N^2{coef_bub[1]: .2}*N+{coef_bub[2]: .2}' )
plt.plot(x, y_ins, label=f'Insert_sort: t={coef_ins[0]: .2}*N^2{coef_ins[1]: .2}*N+{coef_ins[2]: .2}')
plt.plot(x, y_sel, label=f'Select_sort: t={coef_sel[0]: .2}*N^2{coef_sel[1]: .2}*N+{coef_sel[2]: .2}')

plt.xlabel('Number of items')
plt.ylabel('Time of sorts')
plt.grid()
plt.legend()
plt.savefig('Slow_sorts.pdf', format='pdf')
plt.show()
```

```python
log_N = np.log (N)
log_bub= np.log (bub)
log_ins = np.log (ins)
log_sel = np.log (sel)

plt.figure(figsize=(10, 6))

plt.scatter(log_N, log_bub, s=30)
plt.scatter(log_N, log_ins, s=30)
plt.scatter(log_N, log_sel, s=30)

plt.plot(log_N, log_bub, label='Buble_sort')
plt.plot(log_N, log_ins, label='Insert_sort')
plt.plot(log_N, log_sel, label='Select_sort')

plt.xlabel('log(Number of items)')
plt.ylabel('log(Time of sorts)')
plt.grid()
plt.legend()
plt.savefig('Slow_sorts_log.pdf', format='pdf')
plt.show()
```

```python
def is_sorted(mass):
    n=len(mass)
    for i in range (n-1):
        if mass[i]>mass[i+1]:
            return False
    return True

def swap(a,b):
    tmp = a
    a = b
    b = tmp

def bubble_sort (mass):
    n = len(mass)
    for i in range (n):
        for j in range (n-i-1):
            if mass[j]>mass[j+1]:
                swap(mass[j],mass[j+1])

def find_min (mass, start):
    n = len(mass)
    min_index = start
    for i in range(n):
        if mass[i] < mass[min_index]:
            min_index = j
    return min_index

def select_sort (mass):
    n=len(mass)
    for i in range (n):
        index = find_min(mass, i)
        swap(mass[i], mass[index])
```

```python
test_size=[10, 100, 1000, 5000, 10000]
time1 = []
time2 = []
for i in test_size:
    test_mass=[]
    for j in range (i):
        test_mass.append (randint(0, 100000))

    t1 = time()
    bubble_sort(test_mass)
    t2 = time()

    time1.append(t2 - t1)
    print(i, 'Time buble_sort: ', t2 -t1)

for i in test_size:
    test_mass=[]
    for j in range (i):
        test_mass.append (randint(0, 100000))

    t1 = time()
    select_sort(test_mass)
    t2 = time()

    time2.append(t2 - t1)
    print(i, 'Time select_sort: ', t2 -t1)
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(test_size, time1, s=30, label='Buble_python')
plt.scatter(test_size, time2, s=30, label='Select_python')
plt.scatter(N[:5], bub[:5], s=30, label='Buble_cpp')
plt.scatter(N[:5], sel[:5], s=30, label='Select_cpp')

plt.plot(test_size, time1)
plt.plot(test_size, time2)
plt.plot(N[:5], bub[:5])
plt.plot(N[:5], sel[:5])

plt.xlabel('Number of items')
plt.ylabel('Time of sorts')
plt.grid()
plt.legend()
plt.savefig('Comparing_python.pdf', format='pdf')
plt.show()
```
