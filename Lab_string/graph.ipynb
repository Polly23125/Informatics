```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
mass1 = np.array(pd.read_csv("1.csv"))

n=[16, 1024, 16384]
t=[]

for i in range (len(mass1)):
    t.append(mass1[i][-1])

scalar = t[:3]

b_value_short = [t[3], t[6], t[9]]
b_link_short = [t[4], t[7], t[10]]
b_pointer_short = [t[5], t[8], t[11]]

text_value_short = [t[21], t[24], t[27]]
text_link_short = [t[22], t[25], t[28]]
text_pointer_short = [t[23], t[26], t[29]]

b_value_full = [t[12], t[15], t[18]]
b_link_full = [t[13], t[16], t[19]]
b_pointer_full = [t[14], t[17], t[20]]

text_value_full = [t[30], t[33], t[36]]
text_link_full = [t[31], t[34], t[37]]
text_pointer_full = [t[32], t[35], t[38]]
```

```python
plt.scatter(n, b_value_short, s=30)
plt.scatter(n, b_link_short, s=30)
plt.scatter(n, b_pointer_short, s=30)

plt.plot(n, b_value_short, label='Value' )
plt.plot(n, b_link_short, label='Link')
plt.plot(n, b_pointer_short, label='Pointer')

plt.title("Короткая обработка без оптимизаций")
plt.xlabel('Number of items')
plt.ylabel('Time of call')
plt.grid()
plt.legend()
plt.savefig('Block_short.pdf', format='pdf')
plt.show()
```
