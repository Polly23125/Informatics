```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
mass1 = np.array(pd.read_csv("results.csv"))

n=[]
t1=[]
t2=[]
t3=[]
t4=[]
t5=[]

for i in range (len(mass1)):
    n.append(mass1[i][0])
    t1.append(mass1[i][1])
    t2.append(mass1[i][2])
    t3.append(mass1[i][3])
    t4.append(mass1[i][4])
    t5.append(mass1[i][5])

plt.figure(figsize=[6,4])
plt.scatter(n, t1, s=30)
plt.scatter(n, t2, s=30)
plt.scatter(n, t3, s=30)
plt.scatter(n, t4, s=30)
plt.scatter(n, t5, s=30)

plt.plot(n, t1, label='Vector' )
plt.plot(n, t2, label='Subvector add 10')
plt.plot(n, t3, label='Subvector add 100')
plt.plot(n, t4, label='Subvector myltiply 4')
plt.plot(n, t5, label='Subvector myltiply 2')

plt.xlabel('Number of addings')
plt.ylabel('Time of add')
plt.grid()
plt.legend()
plt.savefig('vec_and_sub.pdf', format='pdf')
plt.show()
```
