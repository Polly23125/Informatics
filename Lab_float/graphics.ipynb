```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
mass1 = np.array(pd.read_csv("value_of_pi.csv"))
N1 = []
leib = []
bbp = []
sqrt = []
nil = []
wal = []
tru = []

for i in range (len(mass1)):
    N1.append(int(mass1[i][0]))
    leib.append(float(mass1[i][1]))
    bbp.append(float(mass1[i][2]))
    sqrt.append(float(mass1[i][3]))
    nil.append(float(mass1[i][4]))
    wal.append(float(mass1[i][5]))
    tru.append(3.141592653589793)

N1 = np.array(N1)
leib = np.array(leib)
bbp = np.array(bbp)
sqrt = np.array(sqrt)
nil = np.array(nil)
wal = np.array(wal)
tru = np.array(tru)

plt.figure(figsize=(10, 6))

plt.scatter(N1, leib, s=30, label='Leibniz')
plt.scatter(N1, bbp, s=30, label='BBP')
plt.scatter(N1, nil, s=30, label='Nilakantha')
plt.scatter(N1, wal, s=30, label='Wallis')

plt.plot(N1, leib)
plt.plot(N1, bbp)
plt.plot(N1, nil)
plt.plot(N1, wal)
plt.plot(N1, tru, linewidth = 3.5)


plt.xlabel('Iteration')
plt.ylabel('Value of pi')
plt.grid()
plt.legend()
plt.savefig('Value_of_pi.pdf', format='pdf')
plt.show()
```
